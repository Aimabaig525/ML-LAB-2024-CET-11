```python
def findDog(text):
    return 'dog' in text.lower().split()
print(findDog('Is there a Dog here?'))  
print(findDog('Is there a cat here?')) 
```

```python

def countDog(text):
    count = 0
    for word in text.lower().split():
        if word == 'dog':
            count += 1
    return count

print(countDog('This dog runs faster than the other dog dude!')) # 2
```

```python
def caught_speeding(speed, is_birthday):
    # On birthday, you get 5 mph extra
    if is_birthday:
        speed = speed - 5
    
    if speed <= 60:
        return "No Ticket"
    elif speed >= 61 and speed <= 80:
        return "Small Ticket"
    else:
        return "Big Ticket"
print(caught_speeding(60, False))   
print(caught_speeding(65, False))   
print(caught_speeding(65, True))    
print(caught_speeding(81, False))  
print(caught_speeding(85, True))    
```
